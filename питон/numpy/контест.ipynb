```python
n, m = input().split()
n, m = int(n), int(m)
k = int(input())
field = [[0 for i in range(m)] for j in range(n)]
for i in range(k):
    x, y, d = input().split()
    x, y, d = int(x), int(y), int(d)
    field[x][y]=1
    for j in range(d+1):
        if x-j>=0: field[x-j][y]=1
        if x+j<=n-1: field[x+j][y]=1
        if y-j>=0: field[x][y-j]=1
        if y+j<=m-1: field[x][y+j]=1 
        for z in range(j+1):
            if x-j>=0 and y-z>=0: field[x-j][y-z]=1
            if x-j>=0 and y+z<=m-1: field[x-j][y+z]=1
            if x+j<=n-1 and y-z>=0: field[x+j][y-z]=1
            if x+j<=n-1 and y+z<=m-1: field[x+j][y+z]=1
        for z in range(j+1):
            if x-z>=0 and y-j>=0: field[x-z][y-j]=1
            if x-z>=0 and y+j<=m-1: field[x-z][y+j]=1
            if x+z<=n-1 and y-j>=0: field[x+z][y-j]=1
            if x+z<=n-1 and y+j<=m-1: field[x+z][y+j]=1
ans = 0
for i in range(n):
    for j in range(m):
        if field[i][j]==0: ans+=1
print(ans)
```

```python
import numpy as np

n, m = input().split()
n, m = int(n), int(m)

field = np.array([], dtype=int)

for i in range(n):
    h = np.array(input().split(), dtype=int)
    field = np.append(field, h)

print(len([i for i in field if i<-5]))
print(abs(sum(np.array([i for i in field if i<0]))))
print(max(field))
```
